```python

# Start Ollama in the background
import subprocess
import time

# Start the Ollama server
server = subprocess.Popen(
    ["ollama", "serve"],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL
)

# Wait a few seconds for the server to start
time.sleep(5)

print("Ollama server started")
```
