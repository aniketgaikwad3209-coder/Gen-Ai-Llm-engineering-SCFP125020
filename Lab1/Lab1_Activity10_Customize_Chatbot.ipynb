```python

# Give the chatbot a role
role = "You are a friendly AI and ML tutor."

# Ask a question
question = "What is machine learning?"

# Display the role and question
print("Chatbot Role:", role)
print("My Question:", question)

# Display a sample answer
answer = "Machine learning is a type of AI that learns patterns from data."

print("Chatbot Answer:", answer)
```
