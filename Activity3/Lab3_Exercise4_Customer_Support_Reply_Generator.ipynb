```python

import os
from groq import Groq

# Get the API key
api_key = os.getenv("GROQ_API_KEY")

# Connect to Groq
client = Groq(api_key=api_key)


# Create a function to generate a support reply
def generate_support_reply(customer_message, company_name, max_words):

    # Create the prompt
    prompt = f"""
You are a customer support assistant for {company_name}.

Customer message:
{customer_message}

Instructions:
1. Write a polite and helpful reply.
2. Use a friendly and professional tone.
3. Keep the reply within {max_words} words.
4. Do not make promises that are not supported by the message.
5. Return only the reply.
"""

    # Send the prompt to the AI model
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {"role": "user", "content": prompt}
        ],
        temperature=0
    )

    # Get the generated reply
    reply = response.choices[0].message.content.strip()

    return reply


# Example customer information
customer_message = "My order has not arrived yet. Please help."
company_name = "ABC Store"
max_words = 50


# Run the program
reply = generate_support_reply(
    customer_message,
    company_name,
    max_words
)

# Print the result
print("Customer Message:", customer_message)
print("\nSupport Reply:")
print(reply)
```
