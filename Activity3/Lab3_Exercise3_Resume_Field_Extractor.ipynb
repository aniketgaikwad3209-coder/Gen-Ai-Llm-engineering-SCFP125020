```python

import os
import json
from groq import Groq

# Get the API key
api_key = os.getenv("GROQ_API_KEY")

# Connect to Groq
client = Groq(api_key=api_key)


# Create a function to extract resume details
def extract_resume_fields(resume_text, fields_to_extract):

    # Create the prompt
    prompt = f"""
You are a resume field extractor.

Read the resume and extract only the requested fields.

Resume:
{resume_text}

Fields to extract:
{fields_to_extract}

Rules:
1. Return valid JSON only.
2. Include only the requested fields.
3. If a field is missing, use null.
4. Do not add explanations outside the JSON.
"""

    # Send the prompt to the AI model
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {"role": "user", "content": prompt}
        ],
        temperature=0
    )

    # Get the model's answer
    answer = response.choices[0].message.content.strip()

    # Convert the JSON text into Python data
    return json.loads(answer)


# Example resume text
resume_text = """
Name: Rahul Patil
Education: Bachelor of Computer Science
Skills: Python, Java, SQL
Experience: 1 year as a Junior Developer
"""

# Fields we want to extract
fields_to_extract = ["Name", "Education", "Skills"]

# Run the program
result = extract_resume_fields(resume_text, fields_to_extract)

# Print the result as formatted JSON
print(json.dumps(result, indent=4))
```
