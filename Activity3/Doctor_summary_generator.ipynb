```python

import os
from groq import Groq

# Get the API key
api_key = os.getenv("GROQ_API_KEY")

# Connect to Groq
client = Groq(api_key=api_key)


# Create a function to generate the doctor summary
def generate_doctor_summary(patient_name, patient_notes):

    # Write the prompt for the AI
    prompt = f"""
You are a medical documentation assistant.

Patient Name: {patient_name}

Patient Notes:
{patient_notes}

Create a clean summary using exactly these sections:

Symptoms:
Diagnosis:
Recommendation:

Rules:
1. Use only the information in the notes.
2. Do not invent medical information.
3. If information is missing, write "Not specified".
4. Return only the summary.
"""

    # Send the prompt to the AI model
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {"role": "user", "content": prompt}
        ],
        temperature=0
    )

    # Get the summary from the AI
    summary = response.choices[0].message.content

    return summary


# Example patient information
patient_name = "Ishwar Jadhav"

patient_notes = """
Patient has fever and cough for three days.
Temperature is 101 F.
Doctor suspects a viral infection.
Doctor advised rest, fluids, and follow-up if symptoms worsen.
"""


# Run the program
summary = generate_doctor_summary(patient_name, patient_notes)

# Print the result
print("===== DOCTOR SUMMARY =====")
print("Patient Name:", patient_name)
print(summary)
```
