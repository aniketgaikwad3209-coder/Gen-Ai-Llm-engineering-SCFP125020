```python

import os
from groq import Groq

# Get the API key
api_key = os.getenv("GROQ_API_KEY")

# Connect to Groq
client = Groq(api_key=api_key)


# Create a function to classify customer reviews
def classify_review(review_text):

    # Create the prompt
    prompt = f"""
You are a customer review classifier.

Read the review and choose one category.

Categories:
Positive - The customer is happy.
Negative - The customer is unhappy.
Neutral - The customer has no clear positive or negative opinion.

Customer Review:
{review_text}

Return only one category:
Positive, Negative, or Neutral.
"""

    # Send the review to the AI model
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {"role": "user", "content": prompt}
        ],
        temperature=0
    )

    # Get the category
    category = response.choices[0].message.content.strip()

    return category


# Example customer review
review_text = "The product is excellent and delivery was fast."


# Run the program
category = classify_review(review_text)

# Print the result
print("Customer Review:", review_text)
print("Review Category:", category)
```
