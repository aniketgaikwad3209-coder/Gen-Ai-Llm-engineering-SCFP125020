```python

import os
from groq import Groq

# Get the API key
api_key = os.getenv("GROQ_API_KEY")

# Connect to Groq
client = Groq(api_key=api_key)


# Create a translation function
def translate_text(sentence, target_language, formality):

    # Create the prompt
    prompt = f"""
You are a language translator.

Translate this sentence into {target_language}.

Sentence:
{sentence}

Formality level: {formality}

Instructions:
1. Keep the original meaning.
2. Use the requested formality level.
3. Return only the translated sentence.
"""

    # Send the prompt to the AI model
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {"role": "user", "content": prompt}
        ],
        temperature=0
    )

    # Get the translated sentence
    translation = response.choices[0].message.content.strip()

    return translation


# Example input
sentence = "How are you?"
target_language = "Marathi"
formality = "Formal"


# Run the program
translation = translate_text(
    sentence,
    target_language,
    formality
)

# Display the result
print("Original Sentence:", sentence)
print("Target Language:", target_language)
print("Formality:", formality)
print("Translation:", translation)
```
